```python
# Total missing values
print("Total missing values:", df.isnull().sum().sum())
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
```

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates removed")
```

```python
# Select numerical columns
num = df.select_dtypes(include='number').columns

# Fill missing values with mean
for col in num:
    df[col] = df[col].fillna(df[col].mean())

print("Missing numerical values handled")
```

```python
# Select categorical columns
cat = df.select_dtypes(include='object').columns

# Fill missing values with mode
for col in cat:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing categorical values handled")
```

```python
print(df.isnull().sum())
```

```python
df['popularity_category'] = 'low'

df.loc[df['Rank'] >= 40,'popularity_category'] = 'Medium'
df.loc[df['Rank'] >= 70,'popularity_category'] = 'high'
df.head()
```

```python
top_songs = df.sort_values(
    by='Rank',
    ascending=True
)

top_songs[['Rank','Title',]].head(10)
```

```python
top_10 = df.sort_values(by='Rank').head(10)

plt.figure(figsize=(10, 5))

plt.bar(top_10['Title'], top_10['Rank'])

plt.xticks(rotation=90)

plt.xlabel("Song Name")
plt.ylabel("Rank")

plt.title("Top 10 Spotify Songs")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Energy'], bins=20)

plt.xlabel("Energy")
plt.ylabel("Number of Songs")

plt.title("Distribution of Song Energy")

plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=False,
    cmap='coolwarm'
)

plt.title("Correlation Between Spotify Features")

plt.show()
```
